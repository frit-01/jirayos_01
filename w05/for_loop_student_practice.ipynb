# Challenge — Mini Grade Calculator

เขียนโปรแกรมหนึ่งชุดที่ทำงานครบทุกข้อด้านล่าง

1. แสดงชื่อ คะแนน และเกรดของนักศึกษาแต่ละคน
2. หาคะแนนรวมและค่าเฉลี่ยของทั้งห้อง
3. นับจำนวนผู้ผ่านและไม่ผ่าน
4. นับจำนวนผู้ได้เกรด A, B, C, D และ F

**ข้อกำหนด:** ใช้ `for name, score in students:` เพียงหนึ่ง loop สำหรับการประมวลผลนักศึกษา

**ผลลัพธ์ตัวอย่าง (รูปแบบอาจต่างกันได้)**
```text
นันท์ | Score: 88 | Grade: A
บีม | Score: 72 | Grade: B
ซี | Score: 65 | Grade: C
ดี | Score: 54 | Grade: D
อี | Score: 43 | Grade: F

Class summary
Average score: 64.4
Passed: 4
Failed: 1
A: 1, B: 1, C: 1, D: 1, F: 1
```

````python
students = [
    ("นันท์", 88),
    ("บีม", 72),
    ("ซี", 65),
    ("ดี", 54),
    ("อี", 43)
]

# TODO: กำหนดตัวแปร total, pass_count, fail_count และตัวนับเกรด

# TODO: วนซ้ำ students เพื่อแสดงผลและคำนวณสรุป

# TODO: แสดง Class summary หลังจบ loop
A = 0
B = 0
C = 0
D = 0
F = 0
pass_sd = 0
fail_sd = 0
total = 0
num = len(students)
for student in students:
    name, score= student
    if score >= 80:
        print(f"{name} | Score: {score} | Grade: A")
        total += score
        pass_sd += 1
        A += 1
    elif score >= 70:
        print(f"{name} | Score: {score} | Grade: B")
        total += score
        pass_sd += 1
        B += 1
    elif score >= 60:
        print(f"{name} | Score: {score} | Grade: C")
        total += score
        pass_sd += 1
        C += 1
    elif score >= 50:
        print(f"{name} | Score: {score} | Grade: D")
        total += score
        pass_sd += 1
        D += 1
    else:
        print(f"{name} | Score: {score} | Grade: F")
        total += score
        fail_sd +=1
        F += 1

print("Class summary")
average = total/num
print(f"Average score: {average}")
print(f"Passed: {pass_sd}")
print(f"Failed: {fail_sd}")
print(f"A: {A} B: {B} C: {C} D: {D} F: {F}")
````
