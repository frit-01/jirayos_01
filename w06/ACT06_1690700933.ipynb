# 1. จาก `for` loop สู่ `while` loop

เราใช้ `for` เมื่อทราบชุดข้อมูลหรือจำนวนรอบอย่างชัดเจน เช่น แสดงเลข 1 ถึง 5


````python
for number in range(1, 6):
    print(number)
````

แต่บางปัญหาเรา **ไม่ทราบจำนวนรอบล่วงหน้า** เช่น

- รับรหัสผ่านจนกว่าจะถูกต้อง
- รับคะแนนจนผู้ใช้พิมพ์ `-1` เพื่อจบรายการ
- เล่นเกมตราบใดที่พลังชีวิตยังมากกว่า 0

กรณีเช่นนี้ใช้ `while` เพราะโปรแกรมทำซ้ำ **ขณะที่ (while) เงื่อนไขเป็น `True`**

## รูปแบบพื้นฐาน

```python
ค่าเริ่มต้น
while เงื่อนไข:
    คำสั่งที่ทำซ้ำ
    การเปลี่ยนแปลงค่า
```

Python ตรวจเงื่อนไข *ก่อน* เริ่มแต่ละรอบ: ถ้าเป็น `True` จะทำบล็อกคำสั่ง ถ้าเป็น `False` จะออกจาก loop

# 2. ตัวอย่างแรก: นับเลขด้วย `while`

**ก่อน Run:** ทำนายว่าโปรแกรมจะแสดงเลขอะไร และค่าของ `count` หลังจบ loop เท่าไร


````python
count = 1                 # 1) ค่าเริ่มต้น

while count <= 5:         # 2) เงื่อนไข
    print(count)          # 3) งานที่ทำซ้ำ
    count = count + 1     # 4) เปลี่ยนแปลงค่า

print("จบ loop, count =", count)
````

## Trace table

เติมตารางก่อน Run ตัวอย่างด้านบน

| รอบ | `count` ก่อนตรวจ | `count <= 5` | สิ่งที่พิมพ์ | `count` หลังจบรอบ |
|:--:|:--:|:--:|:--:|:--:|
| 1 | 1 | True | 1 | 2 |
| 2 | 2 |  |  |  |
| 3 | 3 |  |  |  |
| 4 | 4 |  |  |  |
| 5 | 5 |  |  |  |
| ตรวจเพื่อออก | 6 |  | - | - |

> สูตรสำคัญ: **เริ่มค่า → ตรวจเงื่อนไข → ทำงาน → เปลี่ยนค่า → ตรวจใหม่**

# 5. ใช้ `while` ร่วมกับ `if / else`

# 6. รับข้อมูลจนกว่าจะถูกต้อง (Input Validation)

จุดเด่นของ `while` คือรับข้อมูลซ้ำได้จนกว่าจะผ่านกติกา ตัวอย่างรับอายุที่ต้องอยู่ระหว่าง 1 ถึง 120


````python
age = int(input("กรอกอายุ (1-120): "))

while age < 1 or age > 120:
    print("ข้อมูลไม่ถูกต้อง กรุณาลองใหม่")
    age = int(input("กรอกอายุ (1-120): "))

print("บันทึกอายุ", age, "ปี เรียบร้อย")
````

### วิเคราะห์โครงสร้าง

- รับค่าครั้งแรก **ก่อน** เข้า `while`
- ถ้าค่าไม่ถูกต้อง เงื่อนไขเป็น `True` จึงเข้า loop
- ต้องรับค่า `age` ใหม่ใน loop เพื่อให้เงื่อนไขเปลี่ยนได้
- เมื่อค่าถูกต้อง เงื่อนไขเป็น `False` และทำงานต่อหลัง loop

# 7. หยุดตามสัญญาณ: Sentinel Value

บางครั้งผู้ใช้เป็นคนบอกเวลาจบด้วยค่าพิเศษ เรียกว่า **sentinel value** เช่น `-1` หมายถึงจบการกรอกคะแนน ตัวอย่างนี้หาผลรวมและค่าเฉลี่ยโดยไม่รวม `-1`


````python
total = 0
count = 0
score = int(input("กรอกคะแนน (พิมพ์ -1 เพื่อจบ): "))

while score != -1:
    total = total + score
    count = count + 1
    score = int(input("กรอกคะแนน (พิมพ์ -1 เพื่อจบ): "))

if count > 0:
    print("ผลรวม =", total)
    print("ค่าเฉลี่ย =", total / count)
else:
    print("ไม่มีคะแนนที่บันทึก")
````

**คำถามชวนคิด**

- เหตุใดต้องตรวจ `if count > 0` ก่อนหาค่าเฉลี่ย?
- หากไม่รับค่า `score` ใหม่ภายใน loop จะเกิดอะไรขึ้น?

<details><summary>แนวคำตอบ</summary>

ต้องป้องกันการหารด้วยศูนย์เมื่อป้อน `-1` ตั้งแต่ครั้งแรก และหากไม่รับค่าใหม่ `score` จะไม่เปลี่ยน จึงอาจเกิด infinite loop

# 8. `break` และ `continue`

- `break` ออกจาก loop ทันที
- `continue` ข้ามคำสั่งที่เหลือของรอบปัจจุบัน และเริ่มรอบถัดไป

โดยทั่วไปให้เขียนเงื่อนไขของ `while` ให้สื่อจุดจบชัดเจนก่อน แล้วใช้สองคำสั่งนี้เมื่อช่วยให้แก้ปัญหาได้ตรงไปตรงมา


````python
# break: หยุดเมื่อพบจำนวนที่หาร 7 ลงตัวตัวแรก
number = 10
while number <= 30:
    if number % 7 == 0:
        print("พบ", number)
        break
    number = number + 1
````

````python
# continue: แสดงเฉพาะเลขคี่ 1 ถึง 10
number = 0
while number < 10:
    number = number + 1
    if number % 2 == 0:
        continue
    print(number)
````

> **ระวัง:** ตัวอย่าง `continue` เพิ่มค่า `number` ก่อนเสมอ หากเพิ่มค่าหลัง `continue` บางเส้นทางอาจทำให้ตัวนับไม่เปลี่ยนและเกิด infinite loop
