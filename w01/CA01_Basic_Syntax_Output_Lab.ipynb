# 6. Challenge — Mini Calculation

ให้สร้างโปรแกรมคำนวณราคาสินค้า โดยกำหนด

- ราคาสินค้า = 120 บาท
- จำนวน = 4 ชิ้น
    
- ส่วนลด = 10%

โปรแกรมควรคำนวณและแสดง

- ราคารวมก่อนลด
- จำนวนเงินส่วนลด
- ราคาสุทธิ

**ไม่จำเป็นต้องใช้ `input()` ในข้อนี้**

````python
# Challenge
# เขียนโปรแกรมของคุณที่นี่
price = 120
quantity = 4
discount = 0.1
total = price * quantity
discount_amount = total * discount
final_total = total - discount_amount
print(f"Total: {total} \nDiscount: {discount_amount} \nFinal Total: {final_total}")
````
