## 1. เงื่อนไขและ Boolean

Conditional statement คือคำสั่งที่ช่วยให้โปรแกรมตัดสินใจ เช่น **ถ้าคะแนนตั้งแต่ 50 ให้สอบผ่าน**

เงื่อนไขต้องให้ผลเป็น Boolean:

- `True` = จริง
- `False` = เท็จ

ตัวดำเนินการเปรียบเทียบที่ใช้บ่อย: `==`, `!=`, `>`, `<`, `>=`, `<=`

ตัวดำเนินการตรรกะ: `and` (และ), `or` (หรือ), `not` (ไม่)

> `=` ใช้กำหนดค่า ส่วน `==` ใช้เปรียบเทียบค่า

````python
age = 20
is_student = True

print(age >= 18)                 # True
print(age < 18)                  # False
print(age >= 18 and is_student)  # True
print(age < 18 or is_student)    # True


````

# 3. คำสั่ง `if`

ใช้เมื่อต้องการให้คำสั่งทำงาน **เฉพาะเมื่อเงื่อนไขเป็นจริง** ถ้าเงื่อนไขเป็นเท็จ โปรแกรมจะข้ามบล็อก `if`

```python

```

````python
# เพิ่มเงินเดือนเฉพาะผู้ที่มีเงินเดือนต่ำกว่า 15,000 บาท
salary = 12000

if salary < 15000:
    salary = salary + 2000
    print("ได้รับการปรับเงินเดือน 2,000 บาท")

print("เงินเดือนปัจจุบัน:", salary)
print("จบการทำงานของโปรแกรม")
````

ลองเปลี่ยน `salary` เป็น `18000` แล้วรันอีกครั้ง ข้อความการปรับเงินเดือนจะไม่แสดง แต่คำสั่งหลัง `if` ยังทำงานเสมอ

### ตัวอย่าง `if` ร่วมกับ `and`
พนักงานจะได้รับเงินเพิ่มเมื่อเงินเดือนอยู่ระหว่าง 15,000–18,000 บาท **และ** ยอดเงินคงเหลือไม่เกิน 200,000 บาท

````python
salary = 16000
balance = 150000

if 15000 <= salary <= 18000 and balance <= 200000:
    salary += 4500 # salary = salary + 4500
    print("ผ่านเงื่อนไขการปรับเงินเดือน")
else:
    print("ไม่ผ่านเงื่อนไขการปรับเงินเดือน")
    
print("เงินเดือนปัจจุบัน:", salary)
````

# 4. คำสั่ง `if-else`

ใช้เมื่อต้องการเลือกทำงาน **2 ทาง**

- เงื่อนไขเป็น `True` → ทำบล็อก `if`
- เงื่อนไขเป็น `False` → ทำบล็อก `else`

```python
if condition:
    statement_when_true
else:
    statement_when_false
```

ในหนึ่งครั้ง โปรแกรมจะเลือกทำงานเพียงหนึ่งบล็อก

````python
# ตรวจสอบผลสอบ
score = 72

if score >= 50:
    result = "Pass"
else:
    result = "Fail"

print("คะแนน:", score)
print("ผลการเรียน:", result)
````

# 5. คำสั่ง `if-elif-else`

ใช้เมื่อมีทางเลือกมากกว่า 2 ทาง Python จะตรวจเงื่อนไขจากบนลงล่าง และทำงานเพียง **บล็อกแรกที่เป็นจริง**

```python
if condition_1:
    statement_1
elif condition_2:
    statement_2
else:
    statement_when_all_false
```

มี `elif` ได้หลายบล็อก แต่ `else` มีได้ไม่เกินหนึ่งบล็อก และอาจไม่เขียนก็ได้

````python
# ตรวจประเภทของจำนวน
number = 0

if number > 0:
    print("Positive number")
elif number == 0:
    print("Zero")
else:
    print("Negative number")
````

## ลำดับของเงื่อนไขสำคัญมาก

สำหรับการตัดเกรด ควรเรียงคะแนนจากสูงไปต่ำ เช่น `>= 80`, `>= 70`, `>= 60` เพราะเมื่อเงื่อนไขหนึ่งเป็นจริงแล้ว Python จะไม่ตรวจ `elif` ที่เหลือ

````python
score = 76

if score >= 80:
    grade = "A"
elif score >= 70:
    grade = "B"
elif score >= 60:
    grade = "C"
elif score >= 50:
    grade = "D"
else:
    grade = "F"

print(f"คะแนน {score} ได้เกรด {grade}")
````

### ตัวอย่าง: เลือกเมนูแปลงอุณหภูมิ

ใช้ `if-elif-else` เพื่อเลือกการทำงานตามค่าเมนู และมี `else` จัดการกรณีเลือกเมนูไม่ถูกต้อง

````python
menu = 2
celsius = 25.0

if menu == 1:
    fahrenheit = celsius * 1.8 + 32
    print(f"อุณหภูมิฟาเรนไฮต์ = {fahrenheit:.1f}")
elif menu == 2:
    romer = celsius * 21 / 40 + 7.5
    print(f"อุณหภูมิโรเมอร์ = {romer:.1f}")
else:
    print("เลือกเมนูไม่ถูกต้อง")
````

# 6. `if` หลายอัน ต่างจาก `if-elif-else`

`if` ที่เขียนแยกกันจะตรวจทุกเงื่อนไข จึงอาจแสดงผลได้หลายบรรทัด ส่วน `if-elif-else` เลือกได้เพียงหนึ่งทาง

````python
score = 90

# ทั้งสองเงื่อนไขเป็นจริง จึงแสดงผล 2 บรรทัด
if score >= 50:
    print("Pass")

if score >= 80:
    print("Very Good")
````

````python
score = 90

# ทำงานเฉพาะบล็อกแรกที่เป็นจริง
if score >= 80:
    print("Very Good")
elif score >= 50:
    print("Pass")
else:
    print("Fail")
````

## แบบฝึกหัดที่ 5: โจทย์ท้าทาย — ค่าส่งสินค้า

กำหนดน้ำหนักพัสดุ `weight` (กิโลกรัม) และคำนวณ `shipping_fee` ตามตาราง

| น้ำหนัก | ค่าส่ง |
|---|---:|
| มากกว่า 0 ถึง 1 กก. | 40 บาท |
| มากกว่า 1 ถึง 3 กก. | 60 บาท |
| มากกว่า 3 ถึง 5 กก. | 80 บาท |
| มากกว่า 5 กก. | 100 บาท |

หากน้ำหนักน้อยกว่าหรือเท่ากับ 0 ให้แสดง `น้ำหนักไม่ถูกต้อง` มิฉะนั้นให้แสดงค่าส่งในรูปแบบ `ค่าส่ง: ... บาท`

ทดสอบอย่างน้อย: `0`, `0.5`, `1`, `1.1`, `3`, `3.1`, `5`, `5.1`


````python
# แบบฝึกหัดที่ 5
weight = 0
if weight <= 0:
    print("น้ำหนักไม่ถูกต้อง")
    exit()
elif weight <= 1:
    shipping_fee = 40
elif 1 < weight <= 3:
    shipping_fee = 60
elif 3 < weight <= 5:
    shipping_fee = 80
else:
    shipping_fee = 10

print(f"ค่าส่งของ: {shipping_fee:.2f} บาท")
````
