## 1. โหมดเปิดไฟล์ที่ต้องรู้

| โหมด | ความหมาย | ข้อควรระวัง |
|---|---|---|
| `"r"` | อ่านไฟล์ | ไฟล์ต้องมีอยู่แล้ว |
| `"w"` | สร้างไฟล์ใหม่ หรือเขียนทับไฟล์เดิม | ข้อมูลเก่าหายทั้งหมด |
| `"a"` | เพิ่มข้อความต่อท้ายไฟล์ | ข้อมูลเดิมยังอยู่ |

ในทุกตัวอย่างจะใช้ `encoding="utf-8"` เพื่อรองรับข้อความภาษาไทย และใช้ `with` เพื่อปิดไฟล์อัตโนมัติ

## 2. Create: เขียน list ลงไฟล์ด้วย `while`

ในตัวอย่างนี้ `tasks` เป็น tuple จึงใช้ตัวแปร `index` ควบคุมการเข้าถึงสมาชิกทีละตัว

> เมื่อรันซ้ำ cell นี้ โหมด `"w"` จะล้าง `tasks.txt` แล้วสร้างเนื้อหาใหม่เสมอ ซึ่งเหมาะกับการสร้างชุดข้อมูลเริ่มต้น


````python
tasks = ("Self-Learning เรื่อง String", "ทำแบบฝึกหัด while", "ส่งงานวันศุกร์")
index = 0

with open(TASK_FILE, "w", encoding="utf-8") as file:
    while index < len(tasks):
        file.write(tasks[index] + "\n")
        index += 1

print(f"เขียน {len(tasks)} รายการลง {TASK_FILE}")
````

````python
# ตรวจสอบผล โดยใช้ while อ่านกลับทีละบรรทัด
with open(TASK_FILE, "r", encoding="utf-8") as file:
    item_number = 1

    while True:
        line = file.readline()
        if line == "":
            break

        print(f"{item_number}. {line.strip()}")
        item_number += 1
````

## 3. Append: เพิ่มข้อมูลต่อท้ายไฟล์

โหมด `"a"` วางตำแหน่งเขียนไว้ท้ายไฟล์ จึงไม่ลบข้อมูลเดิม ตัวอย่างนี้จำลองข้อมูลใหม่เป็น list แล้วใช้ `while` เพิ่มทีละรายการ


````python
new_tasks = ["ทบทวน if-else", "เตรียมคำถามสำหรับคาบถัดไป"]
index = 0

with open(TASK_FILE, "a", encoding="utf-8") as file:
    while index < len(new_tasks):
        file.write(new_tasks[index] + "\n")
        index += 1

print("เพิ่มรายการใหม่แล้ว")
````

## 4. Update: เหตุใดจึงต้องสร้างไฟล์ชั่วคราว?

Text file เป็นข้อมูลต่อเนื่อง จึงไม่สะดวกที่จะแก้ความยาวของบรรทัดตรงกลางไฟล์ เช่น เปลี่ยน `P002,Pen,5` เป็น `P002,Pen,12`

วิธีที่ปลอดภัยและใช้ได้ทั่วไปคือ

1. เปิดไฟล์เดิมเพื่ออ่าน (`"r"`)
2. เปิดไฟล์ชั่วคราวเพื่อเขียน (`"w"`)
3. อ่านทีละบรรทัดด้วย `while` ถ้าเป็นระเบียนเป้าหมายให้เขียนข้อมูลใหม่ มิฉะนั้นคัดลอกบรรทัดเดิม
4. เมื่อทำครบแล้ว ใช้ `replace()` ให้ไฟล์ชั่วคราวแทนที่ไฟล์เดิม

ตัวอย่างนี้ใช้รูปแบบ `รหัสสินค้า,ชื่อสินค้า,จำนวน`


````python
# สร้างไฟล์สินค้าเริ่มต้น (ใช้เพื่อให้ตัวอย่าง update รันได้เสมอ)
INVENTORY_FILE.write_text(
    "P001,Pencil,20\nP002,Pen,5\nP003,Eraser,12\n",
    encoding="utf-8",
)
print(INVENTORY_FILE.read_text(encoding="utf-8"))
````

````python
# ปรับจำนวนของสินค้า P002 เป็น 12 โดยอ่านและเขียนทีละบรรทัดด้วย while
target_sku = "P002"
new_quantity = 12
temp_file = Path("inventory.tmp")
found = False

with open(INVENTORY_FILE, "r", encoding="utf-8") as source:
    with open(temp_file, "w", encoding="utf-8") as destination:
        while True:
            line = source.readline()

            if line == "":
                break

            parts = line.strip().split(",")

            if len(parts) == 3 and parts[0] == target_sku:
                destination.write(f"{parts[0]},{parts[1]},{new_quantity}\n")
                found = True
            else:
                destination.write(line)

if found:
    temp_file.replace(INVENTORY_FILE)
    print(f"ปรับจำนวนสินค้า {target_sku} แล้ว")
else:
    temp_file.unlink()  # ไม่พบข้อมูล จึงไม่เขียนทับไฟล์เดิม
    print(f"ไม่พบรหัสสินค้า {target_sku}")
````

````python
print("ข้อมูลหลังปรับปรุง:")
print(INVENTORY_FILE.read_text(encoding="utf-8"))
````

| งาน | เงื่อนไขของ `while` |
|---|---|
| เขียนสมาชิกใน list/tuple | `index < len(items)` |
| รับข้อมูลจนผู้ใช้สั่งหยุด | `while True` + sentinel + `break` |
| อ่านไฟล์ทีละบรรทัด | `while True` + `readline()` + EOF + `break` |
