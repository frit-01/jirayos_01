## 1. Quick Review

### f-string
```python
name = "Somchai"
age = 20
print(f"ชื่อ {name} อายุ {age} ปี")
```

### `.format()`
```python
print("ชื่อ {} อายุ {} ปี".format(name, age))
```

### Format Specifier
```python
pi = 3.14159265
print(f"Pi = {pi:.2f}")       # 3.14
print(f"Price = {1500:,}")     # 1,500
print(f"{'Hello':>20}")       # ชิดขวา
print(f"{'Hello':<20}")       # ชิดซ้าย
print(f"{'Hello':^20}")       # กึ่งกลาง
print(f"{'='*5}")              # ทำซ้ำ
```

## 2. โจทย์ 2.1 — แสดงข้อมูลนักศึกษา

```python
name = "Chonnawee Phiphatkul"
student_id = "1690712345"
gpa = 3.7568
credits = 21
```

แสดงผลโดยใช้ **f-string**:

```text
========================================
          Student Report Card           
========================================
Name       : Chonnawee Phiphatkul
Student ID : 1690712345
GPA        : 3.76
Credits    : 21
========================================
```

> **Hint:** `:.2f` ทศนิยม 2 ตำแหน่ง, `:^40` กึ่งกลาง 40 ช่อง


````python
# โจทย์ 2.1
name = "jirayos tippason"
student_id = "1690700933"
gpa = 3.8568
credits = 21

print(f"{'='*40}")
print(f"{'Student Report Card':^40}")
print(f"{'='*40}")
print(f"{'Name:':<10} : {name}")
print(f"{'Student ID':<10} : {student_id}")
print(f"{'GPA:':<10} : {gpa:.2f}")
print(f"{'Credits:':<10} : {credits}")
print(f"{'='*40}")
````

## 3. โจทย์ 2.2 — แสดงตารางราคาสินค้า

```python
item1, price1, qty1 = "Notebook", 45.0, 12
item2, price2, qty2 = "Pen", 15.50, 30
item3, price3, qty3 = "Eraser", 8.75, 25
```

แสดงผลโดยใช้ **f-string**:

```text
----------------------------------------------
| Item           | Price    | Qty  | Total    |
----------------------------------------------
| Notebook       |    45.00 |   12 |   540.00 |
| Pen            |    15.50 |   30 |   465.00 |
| Eraser         |     8.75 |   25 |   218.75 |
----------------------------------------------
|                          Grand Total:        |
|                                   1,223.75   |
----------------------------------------------
```

> **Hint:** `:<15` ชิดซ้าย, `:>8.2f` ชิดขวา, `:,.2f` ใส่ comma


````python
# โจทย์ 2.2
item1, price1, qty1 = "Notebook", 45.0, 12
item2, price2, qty2 = "Pen", 15.50, 30
item3, price3, qty3 = "Eraser", 8.75, 25

print(f"{'-'*51}")
print(f"|{'Item':<17}| {'Price':<8} | {'Qty':<8} |{'Total':<8} |")
print(f"{'-'*51}")
print(f"| {item1:<15} | {price1:<8.2f} | {qty1:<8} | {price1*qty1:<8.2f}|")
print(f"| {item2:<15} | {price2:<8.2f} | {qty2:<8} | {price2*qty2:<8.2f}|")
print(f"| {item3:<15} | {price3:<8.2f} | {qty3:<8} | {price3*qty3:<8.2f}|")
print(f"{'-'*51}")
print(f"| {'Grand Total':<15} | {'':<8} | {'':<8} | {price1*qty1 + price2*qty2 + price3*qty3:<8.2f}|")
print(f"{'-'*51}")
````
