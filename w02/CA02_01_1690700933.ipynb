## 2. โจทย์ 1.1 — รับชื่อและอายุ แล้วแสดงผล

1. รับ **ชื่อ** 2. รับ **อายุ** 3. แสดงผล

```text
กรุณาใส่ชื่อ: Somchai
กรุณาใส่อายุ: 20
สวัสดีคุณ Somchai อายุ 20 ปี
```

> **Hint:** ต้องแปลงอายุเป็น `int` ด้วย `int()`


````python
# เขียนคำตอบโจทย์ 1.1
name = input("กรุณาใส่ชื่อ: ")
age = int((input("กรุณาใส่อายุ: ")))
age += 10
print(f"สวัสดี {name} คุณอายุ {age} ปี")
````

## 3. โจทย์ 1.2 — คำนวณพื้นที่สี่เหลี่ยม

1. รับ **ความกว้าง** (width)
2. รับ **ความยาว** (length)
3. คำนวณ **พื้นที่** = width x length
4. แสดง `type()` ของทุกตัวแปร
5. แสดงผลพื้นที่

```text
กรุณาใส่ความกว้าง: 5.5
กรุณาใส่ความยาว: 3.2
ชนิดข้อมูล width = <class 'float'>
ชนิดข้อมูล length = <class 'float'>
ชนิดข้อมูล area = <class 'float'>
พื้นที่สี่เหลี่ยม = 17.6
```

> **Hint:** ใช้ `float()` แปลงค่า


````python
# เขียนคำตอบโจทย์ 1.2
width = float(input("กรุณาใส่ความกว้าง: "))
height = float(input("กรุณาใส่ความสูง: "))
area = width * height
print(f"ประเภทของ width: {type(width)}")
print(f"ประเภทของ height: {type(height)}")
print(f"ประเภทของ area: {type(area)}")
print(f"พื้นที่ของสี่เหลี่ยมผืนผ้า: {area:.2f} ตาราง")
````

## 4. โจทย์ 1.3 — ตรวจสอบชนิดข้อมูล

```python
a = input("Enter a: ")  # ผู้ใช้ป้อน 10
b = int(a)
c = float(a)
d = str(b + c)
```

**คำถาม:** ตัวแปร `a`, `b`, `c`, `d` มีชนิดข้อมูลและค่าอะไร

| ตัวแปร | type | value |
|--------|------|-------|
| a | | |
| b | | |
| c | | |
| d | | |


````python
# ตรวจสอบโจทย์ 1.3
a = input("Enter a: ")  # ผู้ใช้ป้อน 10
b = int(a)
c = float(a)
d = str(b + c)

print(f"ประเภทของ a: {type(a)} {a}\nประเภทของ b: {type(b)} {b}\nประเภทของ c: {type(c)} {c}\nประเภทของ d: {type(d)} {d}")
````
