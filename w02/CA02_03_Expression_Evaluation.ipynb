## 1. Quick Review

### Operator Precedence (ลำดับสูงไปต่ำ)
1. `()` วงเล็บ
2. `**` ยกกำลัง
3. `+x`, `-x` (unary)
4. `*`, `/`, `//`, `%`
5. `+`, `-`

### Mixed Types
- `int + float` → `float`
- `str + str` → `str` (concatenation)
- `str * int` → `str` (repetition)
- `int(7.9)` → `7` (ตัดทศนิยมทิ้ง)
