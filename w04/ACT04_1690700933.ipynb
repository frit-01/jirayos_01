# 1. Accessing Values in Lists

List คือชุดข้อมูลเรียงลำดับ เขียนด้วยวงเล็บเหลี่ยม `[]` จึงเข้าถึงสมาชิกตามตำแหน่งได้

| รูปแบบ | ความหมาย |
|---|---|
| `items[0]` | สมาชิกตัวแรก |
| `items[-1]` | สมาชิกตัวสุดท้าย |
| `items[1:4]` | สมาชิกตำแหน่ง 1 ถึง 3 |
| `items[:3]` | สมาชิกตั้งแต่ต้นถึงตำแหน่ง 2 |
| `items[::2]` | สมาชิกทุก ๆ 2 ตำแหน่ง |


````python
# index ปกติ, negative index และ slicing
print("วิชาแรก:", subjects[0])
print("วิชาสุดท้าย:", subjects[-1])
print("คะแนนลำดับที่ 3:", scores[2])
print("สามวิชาแรก:", subjects[:3])
print("คะแนนตำแหน่ง 1 ถึง 3:", scores[1:4])
print("คะแนนทุก ๆ 2 ตำแหน่ง:", scores[::2])
````

### ระวัง `IndexError`

หาก list มี `n` สมาชิก index ที่ใช้ได้คือ `0` ถึง `n - 1` การใช้ index นอกช่วงจะเกิด `IndexError` ตรวจว่า list ไม่ว่างด้วย `if items:` ก่อนเข้าถึงสมาชิกแรกเมื่อต้องรองรับข้อมูลว่าง


````python
def first_item_or_none(items):
    # คืนสมาชิกแรก หรือ None ถ้า list ว่าง
    return items[0] if items else None


print(first_item_or_none(subjects))
print(first_item_or_none([]))
````

# 2. Updating Lists

List เป็นข้อมูลแบบ **mutable** จึงเปลี่ยนสมาชิกเดิมและเพิ่มสมาชิกใหม่เข้า list เดิมได้

| คำสั่ง | ผล |
|---|---|
| `items[i] = value` | แก้ไขสมาชิกตำแหน่ง `i` |
| `items.append(value)` | เพิ่ม 1 สมาชิกที่ท้าย list |
| `items.extend(iterable)` | เพิ่มสมาชิกทุกตัวจาก iterable |
| `items.insert(i, value)` | แทรกสมาชิกก่อนตำแหน่ง `i` |
| `items += other` | ต่อสมาชิกจาก iterable เข้า list เดิม |


````python
tasks = ["อ่านบทที่ 1", "ทำแบบฝึกหัด"]
print("เริ่มต้น:", tasks)

tasks[0] = "อ่านบท Lists"
tasks.append("ส่งงาน")
tasks.insert(1, "ดูวิดีโอประกอบ")
tasks.extend(["ทบทวน", "พักผ่อน"])

print("หลังแก้ไข:", tasks)
````

### `append()` ต่างจาก `extend()` อย่างไร?

`append(x)` เพิ่ม `x` เป็นสมาชิก **หนึ่งตัว** แม้ `x` จะเป็น list แต่ `extend(x)` นำสมาชิกแต่ละตัวใน `x` มาเพิ่มทีละตัว


````python
a = [1, 2]
b = [1, 2]
a.append([3, 4])
b.extend([3, 4])

print("append:", a, "จำนวนสมาชิก =", len(a))
print("extend:", b, "จำนวนสมาชิก =", len(b))
````

# 4. Basic List Operations

| รูปแบบ | ความหมาย |
|---|---|
| `a + b`, `a * 2` | รวม list, ทำซ้ำสมาชิก |
| `x in a` | ตรวจว่ามี `x` ใน list หรือไม่ |
| `len(a)` | จำนวนสมาชิก |
| `sum(a)`, `min(a)`, `max(a)` | ผลรวม ต่ำสุด และสูงสุดของข้อมูลตัวเลข |
| `sorted(a)` | คืน list ใหม่ที่เรียงแล้ว ไม่เปลี่ยนต้นฉบับ |
| `a.sort()` | เรียง list เดิม และคืนค่า `None` |
| `a.count(x)`, `a.index(x)` | นับจำนวน / index ที่พบครั้งแรก |


````python
numbers = [12, 5, 19, 5, 8]
more_numbers = [3, 21]

print("รวม list:", numbers + more_numbers)
print("ทำซ้ำ:", ["OK"] * 3)
print("มี 19 หรือไม่:", 19 in numbers)
print("จำนวนสมาชิก:", len(numbers))
print("sum, min, max:", sum(numbers), min(numbers), max(numbers))
print("จำนวน 5:", numbers.count(5))
print("ตำแหน่งของ 19:", numbers.index(19))
print("เรียงมากไปน้อย:", sorted(numbers, reverse=True))
print("ต้นฉบับยังเดิม:", numbers)
````

````python
# sort() แก้ list เดิม และไม่คืน list ใหม่
values = [4, 1, 3]
result = values.sort()
print("values หลัง sort():", values)
print("ค่าที่ sort() คืนกลับ:", result)
````

### ค่าสถิติพื้นฐานจาก List

ค่าเฉลี่ย = `sum(data) / len(data)` จึงต้องตรวจว่า list ไม่ว่างก่อนเสมอ

ค่ามัธยฐาน (*median*) คือค่ากลางหลังเรียงข้อมูล: ถ้าจำนวนข้อมูลคี่ใช้ค่าตรงกลาง; ถ้าเป็นจำนวนคู่ใช้ค่าเฉลี่ยของสองค่ากลาง


````python
values = [5, 2, 8, 1, 9]
ordered = sorted(values)
count = len(ordered)
middle = count // 2
median = ordered[middle] if count % 2 else (ordered[middle - 1] + ordered[middle]) / 2
print("Count : ", count)
print("Sum : ", sum(ordered))
print("Minimum : ", ordered[0])
print("Maximum : ", ordered[-1])
print("Average : ", sum(ordered) / count)
print("Median : ", median)


````

# 5. Tuples: ข้อมูลที่เปลี่ยนไม่ได้

Tuple เขียนด้วยวงเล็บ `()` และเข้าถึงสมาชิกด้วย index/slice เหมือน list แต่เมื่อสร้างแล้ว **ไม่สามารถเปลี่ยน เพิ่ม หรือลบสมาชิกโดยตรงได้**

| List | Tuple |
|---|---|
| `[]` | `()` |
| mutable | immutable |
| เหมาะกับข้อมูลที่เปลี่ยนบ่อย | เหมาะกับ record หรือค่าคงที่ |
| มี `append`, `remove`, `sort` | ไม่มี methods เหล่านี้ |


````python
student = ("1670712345", "Chonnawee", 87.5)

print("รหัส:", student[0])
print("ชื่อ:", student[1])
print("คะแนน:", student[-1])

# unpacking: จำนวนตัวแปรต้องเท่ากับจำนวนสมาชิก
student_id, name, score = student
print(f"{name} ({student_id}) ได้ {score} คะแนน")
````

### ข้อควรระวังการใช้ **Tuple**

````python
# tuple สมาชิกเดียวต้องมี comma
one_subject = ("Python",)
not_a_tuple = ("Python")
print(one_subject, type(one_subject))
print(not_a_tuple, type(not_a_tuple))

# คำสั่งนี้เกิด TypeError เพราะ tuple แก้ไขไม่ได้
# student[2] = 90

coordinate = (10, 20)
moved_coordinate = (coordinate[0] + 5, coordinate[1])
print("ก่อนย้าย:", coordinate, "หลังย้าย:", moved_coordinate)
````

### เมื่อต้อง “แก้ไข” Tuple

ให้สร้าง tuple ใหม่แทนของเดิม หรือหาก tuple อยู่ภายใน list ให้แทนที่ **ทั้ง record** ณ index นั้น ไม่ใช่แก้สมาชิกภายใน tuple


````python
colors_tuple = ("red", "green", "blue")
colors_list = list(colors_tuple)
colors_list.append("yellow")
new_colors_tuple = tuple(colors_list)
print(new_colors_tuple)
````

## โจทย์ 2: การค้นหาและคำนวณค่าสถิติพื้นฐานจาก List

กำหนด
```python
exam_scores = [72, 85, 90, 68, 85, 77, 95, 60]
target_score = 85
```

จงเขียนโปรแกรมเพื่อ
1. ตรวจว่ามี `target_score` หรือไม่ หาตำแหน่งแรก และนับความถี่
2. คำนวณจำนวนข้อมูล ผลรวม ค่าเฉลี่ย ต่ำสุด สูงสุด และพิสัย (`max - min`)
3. แสดงคะแนนเรียงจากมากไปน้อย โดยไม่เปลี่ยน list เดิม
4. รองรับกรณี list ว่าง


````python
# CA05 โจทย์ 2 — ลองทำเอง
exam_scores = [72, 85, 90, 68, 85, 77, 95, 60]
target_score = 85

# TODO: ใช้ in, index, count, len, sum, min, max และ sorted

target_exists = target_score in exam_scores
if target_exists:
    first_index = exam_scores.index(target_score)
    frequency = exam_scores.count(target_score)
    print(f"มี {target_score} อยู่ในรายชื่อคะแนน, ตำแหน่งแรก: {first_index}, ความถี่: {frequency}")
else:
    print(f"ไม่มี {target_score} อยู่ในรายชื่อคะแนน")
    print("รายการคะแนนว่าง")

if exam_scores:
    count = len(exam_scores)
    total = sum(exam_scores)
    average = total / count
    minimum = min(exam_scores)
    maximum = max(exam_scores)
    range_score = maximum - minimum
    sorted_scores = sorted(exam_scores, reverse=True)

    print(f"จำนวนข้อมูล: {count}, ผลรวม: {total}, ค่าเฉลี่ย: {average:.2f}, ต่ำสุด: {minimum}, สูงสุด: {maximum}, พิสัย: {range_score}")
    print("คะแนนเรียงจากมากไปน้อย:", sorted_scores)
else:
    print("รายการคะแนนว่าง")

````

## โจทย์ 3: การเข้าถึงและจัดการข้อมูลใน Tuple

ข้อมูลแต่ละคนเป็น tuple `(student_id, name, score)` และเก็บทุก record ไว้ใน list:
```python
students = [("1690712345", "Chonnawee", 82), ("1660754321", "James", 91), ("1680543234", "Bond", 76)]
```

1. ใช้ unpacking แสดงชื่อและคะแนนของทุกคน
2. ค้นหารหัส `1680543234`
3. เปลี่ยนคะแนนของ `1680543234` เป็น `80` โดยสร้าง tuple ใหม่แล้วแทนที่ใน list
4. สร้าง tuple รายชื่อผู้ผ่าน (คะแนนตั้งแต่ 80)


````python
# CA05 โจทย์ 3 — ลองทำเอง
students = [
    ("1690712345", "James", 82),
    ("1660754321", "James", 91),
    ("1680543234", "Bond", 76),
]

# TODO: ค้นหา, แทนที่ tuple ใหม่ และสร้าง passed_names

for student_id, name, score in students:
    print(f"ชื่อ: {name}, คะแนน: {score}")

target_id = "1680543234"
for i, (student_id, name, score) in enumerate(students):
    if student_id == target_id:
        print(f"พบผู้ใช้ ID {target_id}: ชื่อ {name}, คะแนน {score}")
        students[i] = (student_id, name, 80)
        break


passed_names = tuple(name for _, name, score in students if score >= 80)
print("รายชื่อผู้ผ่าน:", passed_names)
````
