## ข้อ 1 [3 คะแนน] ระบบคิดราคาสินค้าพร้อมส่วนลด

เขียนฟังก์ชัน `calc_price(price, qty)` รับราคาต่อชิ้นและจำนวนชิ้น แล้ว **คืนค่า 3 ค่า** ตามลำดับ คือ `total`, `rate`, `net`

- `total` = ราคา × จำนวน
- `rate` คือเปอร์เซ็นต์ส่วนลดที่ได้ ตามยอดรวม `total`
  - ตั้งแต่ 1000 ขึ้นไป ลด 15%
  - ตั้งแต่ 500 ขึ้นไป ลด 10%
  - ตั้งแต่ 200 ขึ้นไป ลด 5%
  - น้อยกว่า 200 ไม่ลด (0%)
- `net` = ยอดรวมหลังหักส่วนลด

จากนั้นเรียกใช้ฟังก์ชันกับชุดทดสอบ `(120, 5)`, `(50, 2)`, `(300, 4)`, `(250, 1)` และแสดงผลดังตัวอย่าง

**ผลลัพธ์ที่คาดหวัง**
```text
Price 120 x 5 | Total: 600 | Discount: 10% | Net: 540.00
Price 50 x 2 | Total: 100 | Discount: 0% | Net: 100.00
Price 300 x 4 | Total: 1200 | Discount: 15% | Net: 1020.00
Price 250 x 1 | Total: 250 | Discount: 5% | Net: 237.50
```
