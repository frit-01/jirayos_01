## ข้อ 1 [3 คะแนน] ระบบคิดราคาสินค้าพร้อมส่วนลด

เขียนฟังก์ชัน `calc_price(price, qty)` รับราคาต่อชิ้นและจำนวนชิ้น แล้ว **คืนค่า 3 ค่า** ตามลำดับ คือ `total`, `rate`, `net`

- `total` = ราคา × จำนวน
- `rate` คือเปอร์เซ็นต์ส่วนลดที่ได้ ตามยอดรวม `total`
  - ตั้งแต่ 1000 ขึ้นไป ลด 15%
  - ตั้งแต่ 500 ขึ้นไป ลด 10%
  - ตั้งแต่ 200 ขึ้นไป ลด 5%
  - น้อยกว่า 200 ไม่ลด (0%)
- `net` = ยอดรวมหลังหักส่วนลด

จากนั้นเรียกใช้ฟังก์ชันกับชุดทดสอบ `(120, 5)`, `(50, 2)`, `(300, 4)`, `(250, 1)` และแสดงผลดังตัวอย่าง

**ผลลัพธ์ที่คาดหวัง**
```text
Price 120 x 5 | Total: 600 | Discount: 10% | Net: 540.00
Price 50 x 2 | Total: 100 | Discount: 0% | Net: 100.00
Price 300 x 4 | Total: 1200 | Discount: 15% | Net: 1020.00
Price 250 x 1 | Total: 250 | Discount: 5% | Net: 237.50
```


````python
def calc_price(price, qty):
    total = price * qty

    if total >= 1000:
        rate = 15
    elif total >= 500:
        rate = 10
    elif total >= 200:
        rate = 5
    else:
        rate = 0

    net = total - total * rate / 100
    return total, rate, net


tests = [(120, 5), (50, 2), (300, 4), (250, 1)]

for price, qty in tests:
    total, rate, net = calc_price(price, qty)
    print(f"Price {price} x {qty} | Total: {total} | Discount: {rate}% | Net: {net:.2f}")
````

## ข้อ 2 [3 คะแนน] เขียนไฟล์ตัวเลข

เขียนโปรแกรมสร้างไฟล์ `numbers.txt` (โหมด `"w"`) โดยใช้ Loop เขียนเลข 1 ถึง 12 บรรทัดละหนึ่งตัว ในรูปแบบ `เลข,ชนิด`

- ถ้าเป็นเลขคู่ ชนิดคือ `Even` ถ้าเป็นเลขคี่ ชนิดคือ `Odd`
- ถ้าเลขนั้นหารด้วย 5 ลงตัว ให้เติม `,Multiple of 5` ต่อท้ายบรรทัด

**เนื้อหาในไฟล์ที่คาดหวัง**
```text
1,Odd
2,Even
3,Odd
4,Even
5,Odd,Multiple of 5
6,Even
7,Odd
8,Even
9,Odd
10,Even,Multiple of 5
11,Odd
12,Even
```


````python
with open("numbers.txt", "w", encoding="utf-8") as f:
    for n in range(1, 13):
        if n % 2 == 0:
            kind = "Even"
        else:
            kind = "Odd"

        line = f"{n},{kind}"
        if n % 5 == 0:
            line += ",Multiple of 5"

        f.write(line + "\n")

print("Saved numbers.txt")
print(open("numbers.txt", encoding="utf-8").read())
````

## ข้อ 3 [6 คะแนน] สรุปยอดขายพนักงาน

ไฟล์ `sales.txt` เก็บยอดขายของพนักงาน รูปแบบ `รหัส, ชื่อ, ยอดขาย` บรรทัดละหนึ่งคน

```text
S001, Beam, 12500
S002, Mint, 8300
S003, Aom, 4200
S004, Best, 15800
S005, Nan, 6900
S006, Fah, 3100
S007, Ploy, 9750
S008, Tae, 11000
```

อ่านไฟล์ทีละบรรทัดด้วย Loop (ข้ามบรรทัดว่าง) เก็บชื่อและยอดขายลงใน List แล้วหา

1. จำนวนพนักงาน, ยอดขายรวม, ยอดขายเฉลี่ย (ทศนิยม 2 ตำแหน่ง)
2. ยอดขายสูงสุดและต่ำสุด พร้อมชื่อของพนักงานคนนั้น
3. นับจำนวนพนักงานแต่ละระดับ โดยใช้ if-elif-else: ยอดขาย ≥ 10000 เป็น A, ≥ 5000 เป็น B, น้อยกว่านั้นเป็น C

**ผลลัพธ์ที่คาดหวัง**
```text
=== Sales Summary ===
Employees: 8
Total Sales: 71550
Average Sales: 8943.75
Highest: Best (15800)
Lowest: Fah (3100)
Grade A: 3
Grade B: 3
Grade C: 2
```


````python
names = []
sales = []

with open("sales.txt", "r", encoding="utf-8") as f:
    while True:
        line = f.readline()
        if line == "":
            break
        if line.strip() == "":
            continue

        parts = line.strip().split(",")
        names.append(parts[1].strip())
        sales.append(int(parts[2]))

count = len(sales)
total = sum(sales)
average = total / count
high = max(sales)
low = min(sales)
high_name = names[sales.index(high)]
low_name = names[sales.index(low)]

grade_a = 0
grade_b = 0
grade_c = 0
for s in sales:
    if s >= 10000:
        grade_a += 1
    elif s >= 5000:
        grade_b += 1
    else:
        grade_c += 1

print("=== Sales Summary ===")
print(f"Employees: {count}")
print(f"Total Sales: {total}")
print(f"Average Sales: {average:.2f}")
print(f"Highest: {high_name} ({high})")
print(f"Lowest: {low_name} ({low})")
print(f"Grade A: {grade_a}")
print(f"Grade B: {grade_b}")
print(f"Grade C: {grade_c}")
````

## ข้อ 4 [8 คะแนน] ระบบรายงานผลการเรียน

ไฟล์ `students.txt` เก็บคะแนนสอบ 3 ครั้งของนักศึกษา รูปแบบ `รหัส, ชื่อ, คะแนน1, คะแนน2, คะแนน3`

```text
65001, Beam, 80, 75, 90
65002, Mint, 92, 88, 95
65003, Aom, 55, 48, 60
65004, Best, 70, 65, 72
65005, Nan, 40, 52, 45
65006, Fah, 85, 90, 78
```

เขียนโปรแกรมที่ประกอบด้วยฟังก์ชันอย่างน้อย 4 ตัวต่อไปนี้ (ชื่อและพารามิเตอร์กำหนดเองได้ตามเหมาะสม)

| ฟังก์ชัน | หน้าที่ |
| --- | --- |
| `read_students(filename)` | อ่านไฟล์ คืนค่าเป็น List ของชื่อและคะแนน |
| `calc_average(scores)` | คืนค่าเฉลี่ยของคะแนน 3 ครั้ง |
| `get_grade(avg)` | คืนเกรด: ≥ 80 เป็น A, ≥ 70 เป็น B, ≥ 60 เป็น C, ≥ 50 เป็น D, น้อยกว่านั้นเป็น F |
| `write_report(...)` | เขียนรายงานลงไฟล์ `report.txt` |

รายงานต้องมีรายชื่อพร้อมค่าเฉลี่ยและเกรดของแต่ละคน ตามด้วยสรุป ได้แก่ จำนวนนักศึกษา ค่าเฉลี่ยของทั้งห้อง คนที่ได้ค่าเฉลี่ยสูงสุดและต่ำสุด จำนวนผู้ผ่าน (เกรดอื่นที่ไม่ใช่ F) และจำนวนคนในแต่ละเกรด

**เนื้อหา `report.txt` ที่คาดหวัง**
```text
=== Student Report ===
Beam     Avg: 81.67  Grade: A
Mint     Avg: 91.67  Grade: A
Aom      Avg: 54.33  Grade: D
Best     Avg: 69.00  Grade: C
Nan      Avg: 45.67  Grade: F
Fah      Avg: 84.33  Grade: A
=== Summary ===
Total Students: 6
Class Average: 71.11
Highest: Mint (91.67)
Lowest: Nan (45.67)
Passed: 5/6
Grade A: 3
Grade B: 0
Grade C: 1
Grade D: 1
Grade F: 1
```


````python
def read_students(filename):
    students = []
    with open(filename, "r", encoding="utf-8") as f:
        lines = f.readlines()

    for line in lines:
        if line.strip() == "":
            continue
        parts = line.strip().split(",")
        name = parts[1].strip()
        scores = [int(parts[2]), int(parts[3]), int(parts[4])]
        students.append([name, scores])

    return students


def calc_average(scores):
    return sum(scores) / len(scores)


def get_grade(avg):
    if avg >= 80:
        return "A"
    elif avg >= 70:
        return "B"
    elif avg >= 60:
        return "C"
    elif avg >= 50:
        return "D"
    else:
        return "F"


def analyze(students):
    records = []
    for name, scores in students:
        avg = calc_average(scores)
        records.append([name, avg, get_grade(avg)])
    return records


def write_report(records, filename):
    avgs = [r[1] for r in records]
    high_index = avgs.index(max(avgs))
    low_index = avgs.index(min(avgs))
    passed = len([r for r in records if r[2] != "F"])

    with open(filename, "w", encoding="utf-8") as f:
        f.write("=== Student Report ===\n")
        for name, avg, grade in records:
            f.write(f"{name:<8} Avg: {avg:.2f}  Grade: {grade}\n")

        f.write("=== Summary ===\n")
        f.write(f"Total Students: {len(records)}\n")
        f.write(f"Class Average: {sum(avgs) / len(avgs):.2f}\n")
        f.write(f"Highest: {records[high_index][0]} ({avgs[high_index]:.2f})\n")
        f.write(f"Lowest: {records[low_index][0]} ({avgs[low_index]:.2f})\n")
        f.write(f"Passed: {passed}/{len(records)}\n")
        for g in ["A", "B", "C", "D", "F"]:
            count = len([r for r in records if r[2] == g])
            f.write(f"Grade {g}: {count}\n")

    print("Report saved successfully!")


students = read_students("students.txt")
records = analyze(students)
write_report(records, "report.txt")
print(open("report.txt", encoding="utf-8").read())
````
